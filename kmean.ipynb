```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
data=pd.read_csv(r"C:\Users\BHIMA SHANKAR\Downloads\Mall_Customers.csv")
data.head(19)
```

```python
x = data[["Annual Income (k$)","Spending Score (1-100)"]]
x
```

```python
plt.scatter(x["Annual Income (k$)"],x["Spending Score (1-100)"])
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.show()
```

```python
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters = k)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.xlabel("k values")
plt.ylabel("WCSS score")
plt.show()
```

```python
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(x)
    pred = kmeans.predict(x)
    score = silhouette_score(x, pred)
    print(f"for{k}=", score)
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(x)
labels = kmeans.predict(x)
labels
```

```python
data["labels"] = labels
data.head()
```

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
x=data[["Annual Income (k$)","Spending Score (1-100)"]]
scaler=StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
plt.figure(figsize=(7,5))
linkage=linkage(x_scaled)
dendrogram(linkage)
plt.show()
```

```python
plt.figure(figsize=(9, 4))
z = linkage(x_scaled, method='ward')
dendrogram(z)
plt.title('Dendrogram for mall customers')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```

```python
model =AgglomerativeClustering(n_clusters=5)
labels = model.fit_predict(x_scaled)
```

```python
labels
```

```python
data["labels"] = labels
data.head()
```

```python
plt.scatter(x["Annual Income (k$)"], x["Spending Score (1-100)"], c=labels, cmap='viridis')
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.show()
```
