```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data=pd.read_csv(r"C:\Users\BHIMA SHANKAR\Downloads\data_visualization.csv")
data.head(10)
```

```python
plt.scatter(data["Science_Score"],data["Department"])
plt.xlabel("Science Score")
plt.ylabel("Department")
plt.title("Science Score vs Department")
plt.show()
```

```python
import matplotlib.pyplot as plt


avg_scores = data.groupby("Department")["Science_Score"].mean()


plt.bar(avg_scores.index, avg_scores.values)

plt.xlabel("Department")
plt.ylabel("Average Science Score")
plt.title("Average Science Score by Department")

plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

avg_scores = data.groupby("Department")["Science_Score"].mean()

plt.pie(avg_scores.values, labels=avg_scores.index, autopct='%1.1f%%')

plt.title("Average Science Score by Department")
plt.show()
```
